```python
import time
import pandas as pd

spark.conf.set("spark.sql.session.timeZone", "UTC")
Q = """SELECT timestampdiff(SECOND, max(fetched_ts), current_timestamp()) AS lag_s
       FROM bootcamp_students.student_jcdc9919_capstone.positions"""

rows = []
for i in range(30):
    lag = spark.sql(Q).first()["lag_s"]
    rows.append({"sample": i + 1, "lag_s": int(lag)})
    time.sleep(10)

s = pd.DataFrame(rows)
print(f"n={len(s)}  min={s.lag_s.min()}s  median={s.lag_s.median():.0f}s  "
      f"p95={s.lag_s.quantile(0.95):.0f}s  max={s.lag_s.max()}s")
display(s)
```
